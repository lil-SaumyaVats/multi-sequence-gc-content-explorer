```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
print("All libraries loaded")
```

```python
df = pd.read_csv("sequences.csv")
df
```

```python
df["sequence"] = df["sequence"].str.upper()
df["length"] = df["sequence"].str.len()
df["gc_count"] = df["sequence"].str.count("G") + df["sequence"].str.count("C")
df["gc_percent"] = (df["gc_count"] / df["length"]) * 100 
df
```

```python
average_gc = np.mean(df["gc_percent"])
print(f"Average GC Content: {average_gc:.2f}%")
```

```python
plt.bar(df["sequence_id"], df["gc_percent"])
plt.title("GC Content Comparison")
plt.xlabel("DNA Sequences")
plt.ylabel("GC Content(%)")

plt.show()
```

```python
results = df[["sequence_id", "length", "gc_count", "gc_percent"]]

results
```
